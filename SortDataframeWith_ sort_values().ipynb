```python
# reading the csv file
df_exams = pd.read_csv(r'C:\Users\Brigita\Downloads\StudentsPerformance.csv')
df_exams.head()
```

## Sort a dataframe

```python
# sort by one column
df_exams.sort_values('math score')
```

```python
# sort descending by one column
df_exams.sort_values('math score', ascending=False)
```

```python
# sort descending by multiple columns
df_exams.sort_values(['math score', 'reading score'], ascending=False)
```

```python
# sort descending by multiple columns and update dataframe using inplace
df_exams.sort_values(['math score', 'reading score'], ascending=False, inplace=True)
```

```python
# sort descending with a key function  sort_values() with a key function uses lowercase temporarily just to determine 
# the sort order—it never alters the actual text stored inside your DataFrame
# Pandas creates lowercased copies behind the scenes to rank the rows, but it outputs the original capitalized values (group A, group E)
df_exams.sort_values(['race/ethnicity'], ascending=True, key=lambda col:col.str.lower())
```
