```python
import pandas as pd
```

```python
# reading the csv file
df_exams = pd.read_csv(r'C:\Users\Brigita\Downloads\StudentsPerformance.csv')
df_exams.head()
```

## Value Counts

```python
# counting gender elements - number of rows

# .count() method, one might want to suround with int() to get the number without data type
df_exams['gender'].count()
```

```python
# counting gender elements - number of rows

# length function
len(df_exams['gender'])
```

```python
# counting gender elements by category
df_exams['gender'].value_counts()
```

```python
# return the relative frequency (divide all values by the sum of values). precentages
df_exams['gender'].value_counts(normalize=True)
```

```python
# counting "parental level of education" elements by category
df_exams['parental level of education'].value_counts()
```

```python
# return the relative frequency and round to 2 decimals. percentages
df_exams['parental level of education'].value_counts(normalize=True).round(2)
```
