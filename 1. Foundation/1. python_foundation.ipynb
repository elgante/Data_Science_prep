## if staments

if <condition>:
    <code>
elif <condition>:
    <code>
...
else:
    <code>

```python
age = 10

if age >= 18:
    print("You are and adult!")
elif age >= 13:
    print("You are a teen!")
else: 
    print("You are a kid")
    
```

    <code>

```python
countries
```

```python
for c in countries:
    print(c)
```

```python
for index, c in enumerate(countries):
    print(index)
    print(c)
```

```python
my_data = {"name":"Frank", "age":"26"}
my_data
```

```python
for key, value in my_data.items():
    print(key)
    print(value)
```
