```python
# reading the csv file
df_exams = pd.read_csv(r'C:\Users\Brigita\Downloads\StudentsPerformance.csv')
df_exams.head()
```

## Create Index

```python
import numpy as np
import random
```

```python
# creating non repetitive values for the index
new_index = np.arange(0, 1000)
new_index
```

```python
# shuffling the indexes
random.shuffle(new_index)
new_index
```

```python
# create a new column with new index
df_exams['new_index'] = new_index
df_exams
```

## Set Index

```python
# set new_index column as index of the dataframe series instead of 1,2,3,4...
df_exams.set_index('new_index', inplace=True)
df_exams
```

## Sort Index

```python
# sort descending by index
df_exams.sort_index(ascending=False)
```
